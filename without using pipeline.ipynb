# **Pipeline in ML**
Pipeline chains together multiple steps so that the output of each step is used as input to the next step


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/train.csv')
```

```python
df.head()
```

```python
df = df.drop(columns = ['Name','PassengerId','Ticket','Cabin'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['Survived']), df['Survived'], test_size=0.2, random_state=0)
```

```python
X_train
```

```python
from sklearn.impute import SimpleImputer
```

```python
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')
```

```python
X_train_age = si_age.fit_transform(X_train[['Age']])
X_train_embarked = si_embarked.fit_transform(X_train[['Embarked']])

X_test_age = si_age.fit_transform(X_test[['Age']])
X_test_embarked = si_embarked.fit_transform(X_test[['Embarked']])
```

```python
X_train_age, X_train_embarked
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe_sex = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
ohe_embarked = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

X_train_sex = ohe_sex.fit_transform(X_train[['Sex']])
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)

X_test_sex = ohe_sex.fit_transform(X_test[['Sex']])
X_test_embarked = ohe_embarked.fit_transform(X_test_embarked)
```

```python
X_train_sex
```

```python
X_train_rem = X_train.drop(columns=['Age','Sex','Embarked'])
X_test_rem = X_test.drop(columns=['Age','Sex','Embarked'])
```

```python
X_train_transformed = np.concatenate((X_train_rem, X_train_sex, X_train_age,X_train_embarked), axis=1)
X_test_transformed = np.concatenate((X_test_rem, X_test_sex, X_test_age,X_test_embarked), axis=1)
```

```python
X_train_transformed
```

```python
X_train_transformed.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf = DecisionTreeClassifier()
```

```python
clf.fit(X_train_transformed, y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, y_pred)
```

```python
import pickle
```

```python
import os
os.makedirs('models', exist_ok=True)
```

```python
pickle.dump(ohe_sex, open('models/ohe_sex.pkl', 'wb'))
pickle.dump(ohe_embarked, open('models/ohe_embarked.pkl','wb'))
pickle.dump(clf, open('models/clf.pkl', 'wb'))
```
