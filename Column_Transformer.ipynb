# **Advanced way to do using ColumnTranformer**




```python
from sklearn.compose import ColumnTransformer
```

```python
df.sample(5)
```

```python
transformer = ColumnTransformer(transformers=[
    ('trf1',SimpleImputer(),['fever']),
    ('trf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough'] ),
    ('trf3', OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32), ['gender','city'])
], remainder='passthrough')
```

```python
X_train_Column_transform = transformer.fit_transform(X_train)
```

```python
X_train_Column_transform
```

```python
X_train_Column_transform.shape
```

```python
X_test_column_transform  = transformer.fit_transform(X_test)
```
