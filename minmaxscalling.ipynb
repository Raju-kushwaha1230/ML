```python
import pandas as pd
import  matplotlib.pyplot  as plt
from sklearn.preprocessing import MinMaxScaler
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("/kaggle/input/wine-quality-dataset/WineQT.csv", usecols=["Id", "alcohol", "citric acid", "pH"])
```

```python
df
```

```python
sns.kdeplot(df['alcohol'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, y_train, y_test = train_test_split(df.drop(columns=["Id"]), df["Id"], test_size=0.2, random_state=42)
```

```python
X_train.shape,X_test.shape, y_train.shape, y_test.shape
```

```python
plt.scatter(X_train['alcohol'], X_train['citric acid'], c= y_train)
plt.xlabel("alcohol")
plt.ylabel("citric acid")
plt.title("Before Scaling")
plt.show()
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled,columns= X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns= X_test.columns)
```

```python
np.round(X_train_scaled.describe(), 1)
```

```python
np.round(X_train.describe(), 1)
```

```python
fig , (ax1, ax2) = plt.subplots(ncols=2, figsize = (12, 5) )

ax1.scatter(X_train['alcohol'], X_train['citric acid'], c = y_train)
ax1.set_title("Before Scalling")

ax2.scatter(X_train_scaled['alcohol'],X_train_scaled['citric acid'], c= y_train)
ax2.set_title("After Scalling")
```

```python
fig , (ax1, ax2) = plt.subplots(ncols=2, figsize = (12, 5) )

ax1.set_title("Before Scalling")
sns.kdeplot(X_train['alcohol'], ax=ax1)
sns.kdeplot(X_train['citric acid'], ax=ax1)

ax2.set_title("After Scalling")
sns.kdeplot(X_train_scaled['alcohol'], ax=ax2)
sns.kdeplot(X_train_scaled['citric acid'] , ax=ax2)

plt.show()
```
